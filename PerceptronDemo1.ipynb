```python
import numpy as np
import pandas as  pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/campusx-official/100-days-of-deep-learning/main/day3/placement.csv"

df = pd.read_csv(url)

print(df.head())
```

```python
sns.scatterplot(
    data=df,
    x='cgpa',
    y='resume_score',
    hue='placed'
)

plt.show()
```

```python
X = df.iloc[:,0:2]
y = df.iloc[:,-1]
```

```python
from sklearn.linear_model import Perceptron
p = Perceptron()
```

```python
p.fit(X,y)
```

```python
p.coef_
```

```python
p.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X.values, y.values, clf=p, legend=5)
```
